```python
import pandas as pd
from datetime import datetime

df = pd.read_csv("data/raw_data.csv")
df
```

```python
from datetime import datetime, timezone

def tiktok_snowflake_to_datetime(snowflake_id: int) -> datetime:
    # Convert to int if it’s a string
    snowflake = int(snowflake_id)
    
    # TikTok’s Snowflake timestamp is in **milliseconds since Unix epoch**
    # extracted by shifting right 22 bits
    timestamp_ms = snowflake >> 22
    
    # Convert ms → seconds and get UTC datetime
    return datetime.fromtimestamp(timestamp_ms / 1000, tz=timezone.utc)

# Apply to DataFrame column
df["post_date_utc"] = df["video_id"].apply(tiktok_snowflake_to_datetime)
```
