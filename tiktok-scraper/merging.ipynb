# merging.ipynb

Merging the datasets from youtube and tiktok

```python
import pandas as pd
```

```python
df_tiktok = pd.read_csv("data/tiktok_clean_data.csv")
df_tiktok.head(2)
```

```python
df_tiktok.columns
```

```python
# keeping only the relevant columns for the tiktok data
df_tiktok = df_tiktok[["source",
                       "tag",
                       "video_id",
                       "description",
                       "likes",
                       "comments",
                       "shares",
                       "views",
                       "author_follower_count",
                       "author_video_count",
                       "datetime"]]
df_tiktok.columns
```

```python
df_yt = pd.read_csv("data/other_sources/yt_dordash_girl_video_finalAGGR_clean.csv")
df_yt.head(2)
```

```python
df_yt.columns
```

```python
# keeping only the relevant columns for the youtube data
df_yt["source"] = "Youtube"
df_yt = df_yt.drop(labels=["url", "accessed_at", "channel_id", "duration"], axis=1)
df_yt.columns
```

```python
# renaming columns to align with tiktok
rename_dict = {
    'video_id': 'video_id',
    'publish_date': 'datetime',
    'description': 'description',
    'channel_following': 'author_follower_count',
    'like_count': 'likes',
    'comment_count': 'comments',
    'view_count': 'views',
}

df_yt_renamed = df_yt.rename(columns=rename_dict)

# fixing datetime
df_yt_renamed['datetime'] = pd.to_datetime(df_yt_renamed['datetime'], format='%m-%d-%y').dt.date
df_tiktok['datetime'] = pd.to_datetime(df_tiktok['datetime']).dt.date


# combining the data
df_combined = pd.concat([df_tiktok, df_yt_renamed], ignore_index=True)
df_combined
```

```python
# exporting

df_combined.to_csv("data/combined_data.csv")
```
